```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# 1. Lê o arquivo CSV
altura_csv = pd.read_csv("../resultados/dados/altura.csv")
```

```python
# 2. Cria os rótulos do eixo X combinando quantidade e ordem
altura_grouped = altura_csv.groupby(["quantidade", "ordem"], as_index=False).mean()
rotulos = altura_grouped["quantidade"].astype(str) + "\n(" + altura_grouped["ordem"] + ")"
x = np.arange(len(rotulos))
largura = 0.4

# 3. Plota as barras lado a lado com cores personalizadas
fig, ax = plt.subplots(figsize=(10, 6))
barras_rn = ax.bar(x - largura/2, altura_grouped["altura_rn"], largura, label="Árvore Rubro-Negra (RN)", color="#2a9d8f")
barras_abb = ax.bar(x + largura/2, altura_grouped["altura_abb"], largura, label="Árvore Binária de Busca (ABB)", color="#e76f51")


# 4. Ajusta escala logarítmica para visualizar ambas as alturas juntas
ax.set_yscale("log")
ax.set_ylabel("Altura da Árvore (escala log)")
ax.set_xlabel("Quantidade de Elementos e Ordem de Inserção")
ax.set_title("Comparação de Altura: Rubro-Negra vs. ABB")
ax.set_xticks(x)
ax.set_xticklabels(rotulos)
ax.legend()
ax.grid(axis="y", linestyle="--", alpha=0.5)

# 5. Adiciona os valores exatos no topo de cada barra
ax.bar_label(barras_rn, padding=3, fontsize=9)
ax.bar_label(barras_abb, padding=3, fontsize=9, fmt="%.0f")

plt.tight_layout()
plt.show()
```
